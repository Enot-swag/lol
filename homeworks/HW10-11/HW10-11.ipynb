```python
# %% [markdown]
# # HW10-11: Классификация изображений и детекция объектов
# 
# ## Часть A: Классификация (STL10)
# ## Часть B: Детекция (Pascal VOC)

# %% [markdown]
# ### 1. Импорты, seed и устройство

# %%
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset, random_split
from torchvision import datasets, transforms, models
from torchvision.models.detection import fasterrcnn_resnet50_fpn
from torchvision.models.detection.faster_rcnn import FastRCNNPredictor
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image
import os
import csv
import json
from tqdm import tqdm
import random

# %%
# Фиксация seed для воспроизводимости
def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(42)

# %%
# Определение устройства
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Используется устройство: {device}")

# %% [markdown]
# ### 2. Часть A: Датасет для классификации (STL10)

# %%
# Преобразования для STL10
def get_transforms_for_aug():
    return transforms.Compose([
        transforms.RandomHorizontalFlip(),
        transforms.RandomRotation(15),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2),
        transforms.RandomResizedCrop(96, scale=(0.8, 1.0)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])

def get_transforms_base():
    return transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])

def get_transforms_resnet():
    return transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])

# %%
# Загрузка STL10 датасета
def load_stl10_datasets(data_dir='./data'):
    # Загружаем train и test
    train_dataset_full = datasets.STL10(
        root=data_dir, split='train', download=True, transform=None
    )
    test_dataset = datasets.STL10(
        root=data_dir, split='test', download=True, transform=get_transforms_base()
    )
    
    # Разделяем train на train и val (80/20)
    train_size = int(0.8 * len(train_dataset_full))
    val_size = len(train_dataset_full) - train_size
    train_dataset, val_dataset = random_split(
        train_dataset_full, [train_size, val_size], 
        generator=torch.Generator().manual_seed(42)
    )
    
    return train_dataset, val_dataset, test_dataset

# %%
# Загрузка датасетов
train_dataset_full, val_dataset_full, test_dataset_full = load_stl10_datasets()

print(f"Train samples: {len(train_dataset_full)}")
print(f"Val samples: {len(val_dataset_full)}")
print(f"Test samples: {len(test_dataset_full)}")

# %%
# Визуализация sanity-check
def show_samples(dataset, num_samples=5):
    fig, axes = plt.subplots(1, num_samples, figsize=(15, 3))
    indices = np.random.choice(len(dataset), min(num_samples, len(dataset)), replace=False)
    
    for i, idx in enumerate(indices):
        img, label = dataset[idx]
        # Проверяем тип изображения
        if isinstance(img, torch.Tensor):
            img = img.permute(1, 2, 0).numpy()
        elif isinstance(img, Image.Image):
            img = np.array(img)
        
        # Нормализуем для отображения
        img = (img - img.min()) / (img.max() - img.min())
        
        axes[i].imshow(img)
        axes[i].set_title(f'Class: {label}')
        axes[i].axis('off')
    plt.tight_layout()
    plt.show()

# Показываем примеры из оригинального датасета
show_samples(train_dataset_full, num_samples=5)

# %%
# Создание DataLoader для разных экспериментов
def create_dataloaders(train_dataset, val_dataset, test_dataset, batch_size=64):
    train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True, num_workers=2)
    val_loader = DataLoader(val_dataset, batch_size=batch_size, shuffle=False, num_workers=2)
    test_loader = DataLoader(test_dataset, batch_size=batch_size, shuffle=False, num_workers=2)
    return train_loader, val_loader, test_loader

# %%
# Функция для применения трансформаций к датасетам
def apply_transform_to_dataset(dataset, transform):
    """Применяет трансформации к датасету"""
    if isinstance(dataset, Subset):
        dataset.dataset.transform = transform
    else:
        dataset.transform = transform
    return dataset

# %% [markdown]
# ### 3. Часть A: Архитектуры моделей (ИСПРАВЛЕННАЯ ВЕРСИЯ)

# %%
# Простая CNN с адаптивным пулингом - работает с любым размером входа
class SimpleCNN(nn.Module):
    def __init__(self, num_classes=10):
        super(SimpleCNN, self).__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 96 -> 48
            nn.Conv2d(32, 64, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 48 -> 24
            nn.Conv2d(64, 128, kernel_size=3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2, 2),  # 24 -> 12
        )
        # Адаптивный пулинг делает размер не зависящим от входного
        self.adaptive_pool = nn.AdaptiveAvgPool2d((4, 4))
        
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 512),  # 128 * 4 * 4 = 2048
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(512, num_classes)
        )
    
    def forward(self, x):
        x = self.features(x)
        x = self.adaptive_pool(x)
        x = self.classifier(x)
        return x

# %%
# Проверка модели
def test_model_shape(model, input_shape=(2, 3, 96, 96)):
    """Тестирует форму выходных данных модели"""
    model.eval()
    model = model.to(device)
    with torch.no_grad():
        test_input = torch.randn(*input_shape).to(device)
        test_output = model(test_input)
        print(f"Входная форма: {input_shape}")
        print(f"Выходная форма: {test_output.shape}")
        print(f"Ожидаемая форма: (batch_size, num_classes) -> ({input_shape[0]}, 10)")
        if test_output.shape == (input_shape[0], 10):
            print("✓ Тест пройден успешно!")
            return True
        else:
            print("✗ Неверная форма выходных данных!")
            return False
    return False

# %%
# Тестируем SimpleCNN
print("Тестирование SimpleCNN с адаптивным пулингом:")
model_test = SimpleCNN(num_classes=10)
test_model_shape(model_test)

# %% [markdown]
# ### 4. Часть A: Функции обучения и оценки

# %%
def train_one_epoch(model, loader, criterion, optimizer, device):
    model.train()
    running_loss = 0.0
    correct = 0
    total = 0
    
    for inputs, labels in tqdm(loader, desc='Training'):
        inputs, labels = inputs.to(device), labels.to(device)
        
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item() * inputs.size(0)
        _, predicted = torch.max(outputs, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()
    
    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc

def evaluate(model, loader, criterion, device):
    model.eval()
    running_loss = 0.0
    correct = 0
    total = 0
    
    with torch.no_grad():
        for inputs, labels in tqdm(loader, desc='Evaluating'):
            inputs, labels = inputs.to(device), labels.to(device)
            outputs = model(inputs)
            loss = criterion(outputs, labels)
            
            running_loss += loss.item() * inputs.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    
    epoch_loss = running_loss / total
    epoch_acc = correct / total
    return epoch_loss, epoch_acc

# %% [markdown]
# ### 5. Часть A: Подготовка датасетов для экспериментов

# %%
# Подготовка датасетов с разными трансформациями
print("Подготовка датасетов для экспериментов...")

# C1: SimpleCNN без аугментаций
train_dataset_c1 = apply_transform_to_dataset(train_dataset_full, get_transforms_base())
val_dataset_c1 = apply_transform_to_dataset(val_dataset_full, get_transforms_base())
test_dataset_c1 = test_dataset_full

train_loader_c1, val_loader_c1, test_loader_c1 = create_dataloaders(
    train_dataset_c1, val_dataset_c1, test_dataset_c1, batch_size=64
)

# C2: SimpleCNN с аугментациями
train_dataset_c2 = apply_transform_to_dataset(train_dataset_full, get_transforms_for_aug())
val_dataset_c2 = apply_transform_to_dataset(val_dataset_full, get_transforms_base())
test_dataset_c2 = test_dataset_full

train_loader_c2, val_loader_c2, test_loader_c2 = create_dataloaders(
    train_dataset_c2, val_dataset_c2, test_dataset_c2, batch_size=64
)

# C3 и C4: ResNet с resizing
train_dataset_c3 = apply_transform_to_dataset(train_dataset_full, get_transforms_resnet())
val_dataset_c3 = apply_transform_to_dataset(val_dataset_full, get_transforms_resnet())
test_dataset_c3 = test_dataset_full

train_loader_c3, val_loader_c3, test_loader_c3 = create_dataloaders(
    train_dataset_c3, val_dataset_c3, test_dataset_c3, batch_size=32
)

print("Все датасеты подготовлены успешно!")

# %%
# Проверка размеров данных
def check_dataloader_shapes(loader, name):
    try:
        inputs, labels = next(iter(loader))
        print(f"{name}: inputs shape = {inputs.shape}, labels shape = {labels.shape}")
        return inputs.shape[1:]  # возвращаем размерность каналов, высоту, ширину
    except Exception as e:
        print(f"Ошибка при проверке {name}: {e}")
        return None

print("\nПроверка размеров данных в DataLoader:")
shape_c1 = check_dataloader_shapes(train_loader_c1, "C1 (без аугментаций)")
shape_c2 = check_dataloader_shapes(train_loader_c2, "C2 (с аугментациями)")
shape_c3 = check_dataloader_shapes(train_loader_c3, "C3/C4 (ResNet)")

# %% [markdown]
# ### 6. Часть A: Функция для проведения экспериментов

# %%
def run_classification_experiment(model, train_loader, val_loader, test_loader, 
                                 optimizer, criterion, device, num_epochs=10, 
                                 experiment_id=''):
    
    history = {'train_loss': [], 'train_acc': [], 'val_loss': [], 'val_acc': []}
    best_val_acc = 0.0
    
    for epoch in range(num_epochs):
        print(f"\n{'='*50}")
        print(f"Epoch {epoch+1}/{num_epochs}")
        print(f"{'='*50}")
        
        train_loss, train_acc = train_one_epoch(model, train_loader, criterion, optimizer, device)
        val_loss, val_acc = evaluate(model, val_loader, criterion, device)
        
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        history['val_loss'].append(val_loss)
        history['val_acc'].append(val_acc)
        
        print(f"\nTrain Loss: {train_loss:.4f}, Train Acc: {train_acc:.4f}")
        print(f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.4f}")
        
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            os.makedirs('homeworks/HW10-11/artifacts', exist_ok=True)
            torch.save(model.state_dict(), f'homeworks/HW10-11/artifacts/best_model_{experiment_id}.pt')
            print(f"✓ Новая лучшая модель сохранена! Val Acc: {best_val_acc:.4f}")
    
    # Финальная оценка на test
    print(f"\n{'='*50}")
    print("Финальная оценка на тестовом наборе:")
    print(f"{'='*50}")
    test_loss, test_acc = evaluate(model, test_loader, criterion, device)
    print(f"Test Loss: {test_loss:.4f}, Test Acc: {test_acc:.4f}")
    
    return history, best_val_acc, test_acc

# %% [markdown]
# ### 7. Часть A: Эксперимент C1 (SimpleCNN без аугментаций)

# %%
print("\n" + "="*60)
print("ЭКСПЕРИМЕНТ C1: SimpleCNN без аугментаций")
print("="*60)

model_c1 = SimpleCNN(num_classes=10).to(device)
criterion = nn.CrossEntropyLoss()
optimizer_c1 = optim.Adam(model_c1.parameters(), lr=0.001)

# Проверяем модель перед обучением
print("\nПроверка модели C1:")
test_model_shape(model_c1, input_shape=(2, 3, 96, 96))

history_c1, best_val_acc_c1, test_acc_c1 = run_classification_experiment(
    model_c1, train_loader_c1, val_loader_c1, test_loader_c1,
    optimizer_c1, criterion, device, num_epochs=10, experiment_id='C1'
)

# %% [markdown]
# ### 8. Часть A: Эксперимент C2 (SimpleCNN с аугментациями)

# %%
print("\n" + "="*60)
print("ЭКСПЕРИМЕНТ C2: SimpleCNN с аугментациями")
print("="*60)

model_c2 = SimpleCNN(num_classes=10).to(device)
optimizer_c2 = optim.Adam(model_c2.parameters(), lr=0.001)

history_c2, best_val_acc_c2, test_acc_c2 = run_classification_experiment(
    model_c2, train_loader_c2, val_loader_c2, test_loader_c2,
    optimizer_c2, criterion, device, num_epochs=10, experiment_id='C2'
)

# %% [markdown]
# ### 9. Часть A: Эксперимент C3 (ResNet18 head-only)

# %%
print("\n" + "="*60)
print("ЭКСПЕРИМЕНТ C3: ResNet18 head-only (замороженный backbone)")
print("="*60)

model_c3 = models.resnet18(weights='IMAGENET1K_V1')
# Замораживаем все слои
for param in model_c3.parameters():
    param.requires_grad = False
# Заменяем классификационную голову
num_features = model_c3.fc.in_features
model_c3.fc = nn.Linear(num_features, 10)
model_c3 = model_c3.to(device)

optimizer_c3 = optim.Adam(model_c3.fc.parameters(), lr=0.001)

# Проверяем модель перед обучением
print("\nПроверка модели C3:")
with torch.no_grad():
    test_input = torch.randn(2, 3, 224, 224).to(device)
    test_output = model_c3(test_input)
    print(f"Входная форма: (2, 3, 224, 224)")
    print(f"Выходная форма: {test_output.shape}")
    if test_output.shape == (2, 10):
        print("✓ Модель работает корректно")
    else:
        print("✗ Ошибка в модели!")

history_c3, best_val_acc_c3, test_acc_c3 = run_classification_experiment(
    model_c3, train_loader_c3, val_loader_c3, test_loader_c3,
    optimizer_c3, criterion, device, num_epochs=10, experiment_id='C3'
)

# %% [markdown]
# ### 10. Часть A: Эксперимент C4 (ResNet18 fine-tuning)

# %%
print("\n" + "="*60)
print("ЭКСПЕРИМЕНТ C4: ResNet18 fine-tuning (разморожен layer4 + fc)")
print("="*60)

model_c4 = models.resnet18(weights='IMAGENET1K_V1')
# Замораживаем ранние слои
for param in model_c4.parameters():
    param.requires_grad = False
# Размораживаем последний блок и fc
for param in model_c4.layer4.parameters():
    param.requires_grad = True
for param in model_c4.fc.parameters():
    param.requires_grad = True

num_features = model_c4.fc.in_features
model_c4.fc = nn.Linear(num_features, 10)
model_c4 = model_c4.to(device)

optimizer_c4 = optim.Adam([
    {'params': model_c4.layer4.parameters(), 'lr': 0.0001},
    {'params': model_c4.fc.parameters(), 'lr': 0.001}
])

history_c4, best_val_acc_c4, test_acc_c4 = run_classification_experiment(
    model_c4, train_loader_c3, val_loader_c3, test_loader_c3,
    optimizer_c4, criterion, device, num_epochs=10, experiment_id='C4'
)

# %% [markdown]
# ### 11. Часть A: Визуализация результатов

# %%
# Сохраняем лучшую модель
best_val_accuracies = [best_val_acc_c1, best_val_acc_c2, best_val_acc_c3, best_val_acc_c4]
best_model_idx = np.argmax(best_val_accuracies)
best_models = [model_c1, model_c2, model_c3, model_c4]
best_model = best_models[best_model_idx]

os.makedirs('homeworks/HW10-11/artifacts', exist_ok=True)
torch.save(best_model.state_dict(), 'homeworks/HW10-11/artifacts/best_classifier.pt')

# Сохраняем конфиг
config = {
    'dataset': 'STL10',
    'architecture': ['SimpleCNN', 'SimpleCNN', 'ResNet18', 'ResNet18'][best_model_idx],
    'experiment': ['C1', 'C2', 'C3', 'C4'][best_model_idx],
    'transforms': 'base' if best_model_idx < 2 else 'resnet',
    'seed': 42,
    'best_val_accuracy': float(best_val_accuracies[best_model_idx]),
    'test_accuracy': float([test_acc_c1, test_acc_c2, test_acc_c3, test_acc_c4][best_model_idx])
}

with open('homeworks/HW10-11/artifacts/best_classifier_config.json', 'w') as f:
    json.dump(config, f, indent=4)

print(f"\nЛучшая модель сохранена: Эксперимент {['C1', 'C2', 'C3', 'C4'][best_model_idx]}")
print(f"Validation Accuracy: {best_val_accuracies[best_model_idx]:.4f}")
print(f"Test Accuracy: {[test_acc_c1, test_acc_c2, test_acc_c3, test_acc_c4][best_model_idx]:.4f}")

# %%
# Графики для лучшей модели
histories = [history_c1, history_c2, history_c3, history_c4]
best_history = histories[best_model_idx]

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Loss
ax1.plot(best_history['train_loss'], label='Train Loss', linewidth=2)
ax1.plot(best_history['val_loss'], label='Val Loss', linewidth=2)
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Loss')
ax1.set_title('Training and Validation Loss - Best Model')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Accuracy
ax2.plot(best_history['train_acc'], label='Train Accuracy', linewidth=2)
ax2.plot(best_history['val_acc'], label='Val Accuracy', linewidth=2)
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Accuracy')
ax2.set_title('Training and Validation Accuracy - Best Model')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.tight_layout()
os.makedirs('homeworks/HW10-11/artifacts/figures', exist_ok=True)
plt.savefig('homeworks/HW10-11/artifacts/figures/classification_curves_best.png', dpi=150)
plt.show()

# %%
# Сравнение экспериментов
experiments = ['C1\n(No Aug)', 'C2\n(Aug)', 'C3\n(Head-Only)', 'C4\n(Fine-Tune)']
val_accuracies = [best_val_acc_c1, best_val_acc_c2, best_val_acc_c3, best_val_acc_c4]
test_accuracies = [test_acc_c1, test_acc_c2, test_acc_c3, test_acc_c4]

x = np.arange(len(experiments))
width = 0.35

fig, ax = plt.subplots(figsize=(12, 6))
bars1 = ax.bar(x - width/2, val_accuracies, width, label='Validation Accuracy', color='skyblue', alpha=0.8)
bars2 = ax.bar(x + width/2, test_accuracies, width, label='Test Accuracy', color='lightcoral', alpha=0.8)

ax.set_xlabel('Experiment', fontsize=12)
ax.set_ylabel('Accuracy', fontsize=12)
ax.set_title('Comparison of Classification Experiments', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(experiments, fontsize=10)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, axis='y')
ax.set_ylim(0, 1.05)

# Добавляем значения на столбцы
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}',
                   xy=(bar.get_x() + bar.get_width() / 2, height),
                   xytext=(0, 3),
                   textcoords="offset points",
                   ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('homeworks/HW10-11/artifacts/figures/classification_compare.png', dpi=150)
plt.show()

# %%
# Визуализация аугментаций
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
sample_img, _ = train_dataset_full[0]

# Получаем изображение в правильном формате
if isinstance(sample_img, torch.Tensor):
    sample_img = sample_img.permute(1, 2, 0).numpy()
elif isinstance(sample_img, Image.Image):
    sample_img = np.array(sample_img)

# Нормализуем для отображения
sample_img = (sample_img - sample_img.min()) / (sample_img.max() - sample_img.min())
transform_aug = get_transforms_for_aug()

for i in range(5):
    # Оригинал
    axes[0, i].imshow(sample_img)
    axes[0, i].set_title('Original', fontsize=10)
    axes[0, i].axis('off')
    
    # Аугментированная версия
    augmented = transform_aug(Image.fromarray((sample_img * 255).astype(np.uint8)))
    augmented = augmented.permute(1, 2, 0).numpy()
    augmented = (augmented - augmented.min()) / (augmented.max() - augmented.min())
    axes[1, i].imshow(augmented)
    axes[1, i].set_title('Augmented', fontsize=10)
    axes[1, i].axis('off')

plt.suptitle('Data Augmentation Examples', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('homeworks/HW10-11/artifacts/figures/augmentations_preview.png', dpi=150)
plt.show()

# %% [markdown]
# ### 12. Часть B: Детекция объектов (Pascal VOC)

# %%
# Загрузка Pascal VOC для детекции
def load_voc_detection(data_dir='./data'):
    print("Загрузка Pascal VOC датасета...")
    train_dataset = datasets.VOCDetection(
        root=data_dir, year='2012', image_set='train', download=True
    )
    val_dataset = datasets.VOCDetection(
        root=data_dir, year='2012', image_set='val', download=True
    )
    return train_dataset, val_dataset

# %%
train_dataset_voc, val_dataset_voc = load_voc_detection()
print(f"\nTrain samples: {len(train_dataset_voc)}")
print(f"Val samples: {len(val_dataset_voc)}")

# %%
# Визуализация примеров из Pascal VOC
def show_voc_samples(dataset, num_samples=3):
    fig, axes = plt.subplots(1, num_samples, figsize=(15, 5))
    indices = np.random.choice(len(dataset), min(num_samples, len(dataset)), replace=False)
    
    for i, idx in enumerate(indices):
        img, target = dataset[idx]
        if isinstance(img, Image.Image):
            img = np.array(img)
        
        axes[i].imshow(img)
        num_objects = len(target['annotation']['object'])
        axes[i].set_title(f'Objects: {num_objects}', fontsize=12)
        axes[i].axis('off')
    
    plt.suptitle('Pascal VOC Dataset Examples', fontsize=14, fontweight='bold')
    plt.tight_layout()
    plt.show()

show_voc_samples(train_dataset_voc)

# %%
# Загрузка предобученной модели Faster R-CNN
def get_fasterrcnn_model(num_classes=21):  # 20 классов VOC + background
    print("\nЗагрузка предобученной модели Faster R-CNN...")
    model = fasterrcnn_resnet50_fpn(weights='DEFAULT')
    in_features = model.roi_heads.box_predictor.cls_score.in_features
    model.roi_heads.box_predictor = FastRCNNPredictor(in_features, num_classes)
    model = model.to(device)
    model.eval()
    print("Модель загружена и переведена в режим оценки")
    return model

model_detection = get_fasterrcnn_model()

# %%
# Функция для преобразования данных для детекции
def transform_voc_image(img):
    transform = transforms.Compose([
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])
    return transform(img).unsqueeze(0).to(device)

# %%
# Визуализация предсказаний детекции
def visualize_detections(img, predictions, ground_truth=None, score_threshold=0.3, save_path=None):
    fig, ax = plt.subplots(1, 1, figsize=(12, 8))
    img_np = np.array(img) if isinstance(img, Image.Image) else img
    ax.imshow(img_np)
    
    # Предсказания
    boxes = predictions[0]['boxes'].cpu().detach().numpy()
    scores = predictions[0]['scores'].cpu().detach().numpy()
    labels = predictions[0]['labels'].cpu().detach().numpy()
    
    for box, score, label in zip(boxes, scores, labels):
        if score >= score_threshold:
            x1, y1, x2, y2 = box
            rect = patches.Rectangle((x1, y1), x2-x1, y2-y1, 
                                    linewidth=2, edgecolor='red', facecolor='none')
            ax.add_patch(rect)
            ax.text(x1, y1-5, f'{label}: {score:.2f}', 
                   color='red', fontsize=8, bbox=dict(facecolor='white', alpha=0.7, edgecolor='none'))
    
    # Ground truth (если есть)
    if ground_truth:
        for obj in ground_truth['annotation']['object']:
            bbox = obj['bndbox']
            x1, y1, x2, y2 = int(bbox['xmin']), int(bbox['ymin']), int(bbox['xmax']), int(bbox['ymax'])
            rect = patches.Rectangle((x1, y1), x2-x1, y2-y1, 
                                    linewidth=2, edgecolor='green', facecolor='none', linestyle='--')
            ax.add_patch(rect)
    
    ax.set_title(f'Detection Results (threshold={score_threshold})', fontsize=12, fontweight='bold')
    ax.axis('off')
    plt.tight_layout()
    if save_path:
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        plt.close()
    else:
        plt.show()
    return fig

# %%
# Функция для расчета метрик детекции
def compute_detection_metrics(predictions, ground_truth, iou_threshold=0.5, score_threshold=0.3):
    true_positives = 0
    false_positives = 0
    false_negatives = 0
    total_iou = 0
    matched_predictions = 0
    
    for pred, gt in zip(predictions, ground_truth):
        pred_boxes = pred['boxes'].cpu().detach().numpy()
        pred_scores = pred['scores'].cpu().detach().numpy()
        pred_labels = pred['labels'].cpu().detach().numpy()
        
        # Фильтруем по порогу уверенности
        valid_preds = pred_scores >= score_threshold
        pred_boxes = pred_boxes[valid_preds]
        pred_scores = pred_scores[valid_preds]
        pred_labels = pred_labels[valid_preds]
        
        # Получаем GT боксы
        gt_boxes = []
        gt_labels = []
        for obj in gt['annotation']['object']:
            bbox = obj['bndbox']
            gt_boxes.append([int(bbox['xmin']), int(bbox['ymin']), 
                            int(bbox['xmax']), int(bbox['ymax'])])
            gt_labels.append(obj['name'])
        
        gt_boxes = np.array(gt_boxes) if gt_boxes else np.array([])
        
        # Сопоставление предсказаний с GT
        matched_gt = set()
        
        for pred_box in pred_boxes:
            if len(gt_boxes) == 0:
                false_positives += 1
                continue
            
            best_iou = 0
            best_gt_idx = -1
            
            for gt_idx, gt_box in enumerate(gt_boxes):
                if gt_idx in matched_gt:
                    continue
                
                # Вычисляем IoU
                intersection_x = max(0, min(pred_box[2], gt_box[2]) - max(pred_box[0], gt_box[0]))
                intersection_y = max(0, min(pred_box[3], gt_box[3]) - max(pred_box[1], gt_box[1]))
                intersection = intersection_x * intersection_y
                
                area_pred = (pred_box[2] - pred_box[0]) * (pred_box[3] - pred_box[1])
                area_gt = (gt_box[2] - gt_box[0]) * (gt_box[3] - gt_box[1])
                union = area_pred + area_gt - intersection
                iou = intersection / union if union > 0 else 0
                
                if iou > best_iou:
                    best_iou = iou
                    best_gt_idx = gt_idx
            
            if best_iou >= iou_threshold and best_gt_idx not in matched_gt:
                true_positives += 1
                matched_gt.add(best_gt_idx)
                total_iou += best_iou
                matched_predictions += 1
            else:
                false_positives += 1
        
        false_negatives += len(gt_boxes) - len(matched_gt)
    
    precision = true_positives / (true_positives + false_positives) if (true_positives + false_positives) > 0 else 0
    recall = true_positives / (true_positives + false_negatives) if (true_positives + false_negatives) > 0 else 0
    mean_iou = total_iou / matched_predictions if matched_predictions > 0 else 0
    
    return precision, recall, mean_iou

# %%
# Проведение экспериментов V1 и V2
def run_detection_experiment(score_threshold, experiment_id):
    print(f"\n{'='*50}")
    print(f"ЭКСПЕРИМЕНТ {experiment_id}: score_threshold = {score_threshold}")
    print(f"{'='*50}")
    
    # Выбираем несколько изображений из validation set
    num_samples = 10
    indices = np.random.choice(len(val_dataset_voc), min(num_samples, len(val_dataset_voc)), replace=False)
    
    all_predictions = []
    all_ground_truth = []
    
    print(f"Обработка {len(indices)} изображений...")
    
    for i, idx in enumerate(indices):
        img, target = val_dataset_voc[idx]
        img_tensor = transform_voc_image(img)
        
        with torch.no_grad():
            predictions = model_detection(img_tensor)
        
        all_predictions.append(predictions[0])
        all_ground_truth.append(target)
        
        # Визуализируем первые 2 изображения
        if i < 2:
            save_path = f'homeworks/HW10-11/artifacts/figures/detection_examples_{experiment_id}_{i+1}.png'
            visualize_detections(img, predictions, target, score_threshold, save_path)
            print(f"  Визуализация {i+1} сохранена")
    
    # Вычисляем метрики
    print("\nВычисление метрик...")
    precision, recall, mean_iou = compute_detection_metrics(
        all_predictions, all_ground_truth, iou_threshold=0.5, score_threshold=score_threshold
    )
    
    print(f"\nРезультаты для {experiment_id}:")
    print(f"  Precision: {precision:.4f}")
    print(f"  Recall: {recall:.4f}")
    print(f"  Mean IoU: {mean_iou:.4f}")
    
    return precision, recall, mean_iou

# %%
# Эксперимент V1: score_threshold = 0.3
precision_v1, recall_v1, mean_iou_v1 = run_detection_experiment(0.3, 'V1')

# %%
# Эксперимент V2: score_threshold = 0.7
precision_v2, recall_v2, mean_iou_v2 = run_detection_experiment(0.7, 'V2')

# %% [markdown]
# ### 13. Визуализация метрик детекции

# %%
# Создание графика сравнения метрик детекции
fig, ax = plt.subplots(figsize=(10, 6))

metrics = ['Precision', 'Recall', 'Mean IoU']
v1_values = [precision_v1, recall_v1, mean_iou_v1]
v2_values = [precision_v2, recall_v2, mean_iou_v2]

x = np.arange(len(metrics))
width = 0.35

bars1 = ax.bar(x - width/2, v1_values, width, label='V1 (threshold=0.3)', color='steelblue', alpha=0.8)
bars2 = ax.bar(x + width/2, v2_values, width, label='V2 (threshold=0.7)', color='darkorange', alpha=0.8)

ax.set_ylabel('Score', fontsize=12)
ax.set_title('Detection Metrics Comparison', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(metrics, fontsize=11)
ax.legend(fontsize=10)
ax.grid(True, alpha=0.3, axis='y')
ax.set_ylim(0, 1.05)

# Добавляем значения на столбцы
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        ax.annotate(f'{height:.3f}',
                   xy=(bar.get_x() + bar.get_width() / 2, height),
                   xytext=(0, 3),
                   textcoords="offset points",
                   ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.savefig('homeworks/HW10-11/artifacts/figures/detection_metrics.png', dpi=150)
plt.show()

# %% [markdown]
# ### 14. Сохранение результатов в runs.csv

# %%
# Создание runs.csv
results = [
    {
        'experiment_id': 'C1',
        'task': 'classification',
        'dataset': 'STL10',
        'seed': 42,
        'model_summary': 'SimpleCNN_Adaptive',
        'optimizer': 'Adam',
        'lr': 0.001,
        'epochs_trained': 10,
        'best_val_accuracy': best_val_acc_c1,
        'test_accuracy': test_acc_c1,
        'precision': None,
        'recall': None,
        'mean_iou': None,
        'notes': 'без аугментаций'
    },
    {
        'experiment_id': 'C2',
        'task': 'classification',
        'dataset': 'STL10',
        'seed': 42,
        'model_summary': 'SimpleCNN_Adaptive',
        'optimizer': 'Adam',
        'lr': 0.001,
        'epochs_trained': 10,
        'best_val_accuracy': best_val_acc_c2,
        'test_accuracy': test_acc_c2,
        'precision': None,
        'recall': None,
        'mean_iou': None,
        'notes': 'с аугментациями'
    },
    {
        'experiment_id': 'C3',
        'task': 'classification',
        'dataset': 'STL10',
        'seed': 42,
        'model_summary': 'ResNet18_head_only',
        'optimizer': 'Adam',
        'lr': 0.001,
        'epochs_trained': 10,
        'best_val_accuracy': best_val_acc_c3,
        'test_accuracy': test_acc_c3,
        'precision': None,
        'recall': None,
        'mean_iou': None,
        'notes': 'замороженный backbone'
    },
    {
        'experiment_id': 'C4',
        'task': 'classification',
        'dataset': 'STL10',
        'seed': 42,
        'model_summary': 'ResNet18_finetune',
        'optimizer': 'Adam',
        'lr': 0.001,
        'epochs_trained': 10,
        'best_val_accuracy': best_val_acc_c4,
        'test_accuracy': test_acc_c4,
        'precision': None,
        'recall': None,
        'mean_iou': None,
        'notes': 'fine-tuning layer4 + fc'
    },
    {
        'experiment_id': 'V1',
        'task': 'detection',
        'dataset': 'PascalVOC',
        'seed': 42,
        'model_summary': 'FasterRCNN_ResNet50_FPN',
        'optimizer': None,
        'lr': None,
        'epochs_trained': 0,
        'best_val_accuracy': None,
        'test_accuracy': None,
        'precision': precision_v1,
        'recall': recall_v1,
        'mean_iou': mean_iou_v1,
        'notes': 'score_threshold=0.3'
    },
    {
        'experiment_id': 'V2',
        'task': 'detection',
        'dataset': 'PascalVOC',
        'seed': 42,
        'model_summary': 'FasterRCNN_ResNet50_FPN',
        'optimizer': None,
        'lr': None,
        'epochs_trained': 0,
        'best_val_accuracy': None,
        'test_accuracy': None,
        'precision': precision_v2,
        'recall': recall_v2,
        'mean_iou': mean_iou_v2,
        'notes': 'score_threshold=0.7'
    }
]

# Сохраняем в CSV
os.makedirs('homeworks/HW10-11/artifacts', exist_ok=True)
with open('homeworks/HW10-11/artifacts/runs.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=results[0].keys())
    writer.writeheader()
    writer.writerows(results)

print("\n" + "="*50)
print("РЕЗУЛЬТАТЫ СОХРАНЕНЫ")
print("="*50)
print(f"✓ runs.csv сохранен в homeworks/HW10-11/artifacts/")
print(f"✓ Лучшая модель классификации сохранена")
print(f"✓ Все визуализации сохранены в artifacts/figures/")

# %%
print("\n" + "="*50)
print("ИТОГОВЫЕ РЕЗУЛЬТАТЫ")
print("="*50)
print("\nКлассификация (лучшая валидационная точность):")
print(f"  C1: {best_val_acc_c1:.4f}")
print(f"  C2: {best_val_acc_c2:.4f}")
print(f"  C3: {best_val_acc_c3:.4f}")
print(f"  C4: {best_val_acc_c4:.4f}")
print(f"\nЛучший эксперимент: C{best_model_idx+1} с точностью {best_val_accuracies[best_model_idx]:.4f}")
print(f"Тестовая точность лучшей модели: {[test_acc_c1, test_acc_c2, test_acc_c3, test_acc_c4][best_model_idx]:.4f}")

print("\nДетекция:")
print(f"  V1 (threshold=0.3): Precision={precision_v1:.4f}, Recall={recall_v1:.4f}, mIoU={mean_iou_v1:.4f}")
print(f"  V2 (threshold=0.7): Precision={precision_v2:.4f}, Recall={recall_v2:.4f}, mIoU={mean_iou_v2:.4f}")
```
